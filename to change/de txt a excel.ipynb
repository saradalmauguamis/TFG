```python
txt_path = "stop_times.txt"
chunk_size = 800_000  # Màxim files per Excel
df = pd.read_csv(txt_path, low_memory=False)

num_chunks = (len(df) // chunk_size) + 1

for i in range(num_chunks):
    start = i * chunk_size
    end = (i + 1) * chunk_size
    chunk = df.iloc[start:end]
    xlsx_path = f"stop_times_part{i+1}.xlsx"
    chunk.to_excel(xlsx_path, index=False)
    print(f"Guardat: {xlsx_path} ({len(chunk)} files)")
```
