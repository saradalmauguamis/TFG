```python
# Llistar "stop_id - stop_name" ordenat per stop_id per a les línies 1 a 5
import os, csv
from typing import Dict, Iterable, List

BASE = r"C:\\Users\\Sarad\\Escritorio\\Mates\\Cursos\\Curs 2025-2026 (3r + 4t)\\TFG\\gtfs"
STOPS_FILE = os.path.join(BASE, "stops.txt")


def sniff_dialect(file_path: str) -> csv.Dialect:
    try:
        with open(file_path, 'r', encoding='utf-8-sig', newline='') as f:
            sample = f.read(65536)
        return csv.Sniffer().sniff(sample, delimiters=",;\t")
    except Exception:
        class _D(csv.Dialect):
            delimiter = ','
            quotechar = '"'
            doublequote = True
            skipinitialspace = False
            lineterminator = '\n'
            quoting = csv.QUOTE_MINIMAL
        return _D


def read_dict_rows(file_path: str) -> Iterable[Dict[str, str]]:
    dialect = sniff_dialect(file_path)
    with open(file_path, 'r', encoding='utf-8-sig', newline='') as f:
        reader = csv.DictReader(f, dialect=dialect)
        if reader.fieldnames is None:
            raise RuntimeError(f"El fitxer {os.path.basename(file_path)} no té capçalera (header).")
        fn_map = {name: name.lower().strip() for name in reader.fieldnames}
        for row in reader:
            out = {}
            for k, v in row.items():
                lk = fn_map.get(k, k).lower()
                sv = (v if v is not None else '').strip()
                out[lk] = sv
            yield out


# Carrega stops i prepara llistes per prefixos de línia
prefixos = ["1.1", "1.2", "1.3", "1.4", "1.5", "1.9"]
llistes: Dict[str, List[Dict[str, str]]] = {p: [] for p in prefixos}

if not os.path.exists(STOPS_FILE):
    print("No s'ha trobat:", STOPS_FILE)
else:
    for r in read_dict_rows(STOPS_FILE):
        sid = r.get('stop_id', '')
        sname = r.get('stop_name', '') or '(sense nom)'
        for p in prefixos:
            if sid.startswith(p):
                llistes[p].append({'stop_id': sid, 'stop_name': sname})
                break

    # Imprimeix per cada línia, ordenat per stop_id
    for p in prefixos:
        numero_linia = p.split('.')[-1]
        print(f"\nLínia {numero_linia}")
        items = sorted(llistes[p], key=lambda x: x['stop_id'])
        if not items:
            print("(Sense resultats)")
            continue
        for it in items:
            print(f"{it['stop_id']} - {it['stop_name']}")
```
